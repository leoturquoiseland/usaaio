```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
Q = np.array([[2, 1], [1, 3]])
c = np.array([1, 2])
A = np.array([[1, 1]])
b = 3
```

```python
def lang(x, v):
    return 1/2 * x.T @ Q @ x + c.T @ x + v.T @ (A @ x - b)
```

```python
arr = np.block([[Q, A.T], [A, 0]])
```

```python
answor = np.array([-c[0], -c[1], b])
```

```python
np.shape(answor)
```

```python
sol = np.linalg.solve(arr, answor)
```

```python
xstar = sol[:2]
```

```python
from scipy.optimize import minimize
```

```python
def obj(x):
    return 0.5 * x @ Q @ x + c @ x
```

```python
constraint = {'type': 'eq', 'fun' : lambda x : A @ x - b}
```

```python
res = minimize(obj, x0= np.array([0 , 0]), constraints = constraint)
print(res.x)
```
