```python
import numpy as np
import matplotlib as plt
import scipy as sp
```

```python
from scipy.optimize import minimize
```

```python
def f(x):
    return 5 * x ** 3 - 10 * x ** 2 + 3 * x - 5
```

```python
def df(x, h = 1e-5):
    return (f(x+h) - f(x-h))/(2 * h)
df(1.0)
```

```python
from scipy.optimize import fsolve
```

```python
root1 = fsolve(df, x0 = 0)
root1
```

```python
root2 = fsolve(df, x0 = 2)
root2
```

```python
def d2f(x, h = 1e-5):
    return(df(x+h) - df(x-h))/(2*h)

d2f(root1)
```

```python
d2f(root2)
```

```python
from scipy.integrate import quad
```

```python
def f(x):
    return x ** 2
```

```python
def trapezodial(f, a, b, n):
    x = np.linspace(a,b, n + 1)
    y = f(x)
    h = (a + b)/n
    coeffs = np.ones_like(y)
    coeffs[1 : -1] = 2
    return (h / 2) * np.sum(y * coeffs)

res = trapezodial(f, 0, 1, 1000000)
res
```

```python
def simpson(f, a, b, n):
    if n % 2 == 1:
        return
    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n
    coeffs = np.ones_like(y)
    coeffs[1 : -1 : 2] = 4
    coeffs[2 : -1 : 2] = 2
    return (h / 3) * np.sum(y * coeffs)

res = simpson(f, 0, 1, 4)
res
```

```python
def integrand(x):
    return x * (1/np.sqrt(2 * np.pi)) * np.exp(-x ** 2 / 2)

res = trapezodial(integrand, -5, 5, 1000)
res
```
