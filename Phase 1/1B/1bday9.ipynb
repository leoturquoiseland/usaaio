```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
from scipy.integrate import quad
from scipy.special import beta
from scipy.integrate import simpson
```

```python
def gaussian2d(x, y):
    return (1 / (2 * np.pi)) * np.exp(-(x ** 2 + y ** 2) / 2)

x = np.linspace(-5, 5, 101)
y = np.linspace(-5, 5, 101)
X, Y = np.meshgrid(x, y)
Z = gaussian2d(X, Y)
```

```python
inner = simpson(Z, x=x, axis = 1)
result = simpson(inner, x = y)
```

```python
def betadistribution(x, alphaa, betaa):
    return (((x ** (alphaa - 1)) * ((1 - x) ** (betaa - 1)))/(beta(alphaa, betaa)))
```
