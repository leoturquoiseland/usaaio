```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
def vv(x):
    return np.array([x[0] ** 2, x[0] * x[1], x[1] ** 2])
```

```python
def jacob(f, x, h):
    xp = x.copy()
    xm = x.copy()
    resa = []
    for i in range(2):
        xp[i] += h
        xm[i] -= h
        fp = vv(xp)
        fm = vv(xm)
        resa.append((fp - fm)/(2 * h))
        xp[i] -= h
        xm[i] += h
    return np.array(resa).T
    
x = np.array([1.0, 2.0])
jacob(vv, x, 1e-6)
```

```python
def softmax(z):
    zshift = z - np.max(z)
    expz = np.exp(zshift)
    sig = expz/np.sum(expz)
    return sig

z = np.array([1.0, 2.0, 0.5])
np.sum(softmax(z))
```

```python
def smjacob(z):
    sig = softmax(z)
    j = np.diag(sig) - np.outer(sig, sig)
    return j

smjacob(z)
```
