```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
np.random.seed(33550336)
rng = np.random.default_rng()
```

```python
flips = []
res = []
cums = []
n = 10000
for i in range(n):
    x = rng.random()
    flips.append(i + 1)
    if x >= 0.5:
        res.append(1)
    else:
        res.append(0)
    cums.append(np.sum(res)/(i + 1))

plt.plot(cums)
plt.axhline(y = 0.5)
```

```python
res = []
cums = []
n = 30000
for i in range(n):
    x = rng.random()
    if x <= 1/6:
        res.append(1)
    elif 1/6 < x <= 2/6:
        res.append(2)
    elif 2/6 < x <= 1/2:
        res.append(3)
    elif 1/2 < x <= 2/3:
        res.append(4)
    elif 2/3 < x <= 5/6:
        res.append(5)
    else:
        res.append(6)
    cums.append(np.sum(res)/(i + 1))

plt.plot(cums)
plt.axhline(y = 3.5)
```
