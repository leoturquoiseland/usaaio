```python
import numpy as numpy
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(33550336)
```

```python
mu = np.array([0, 0])
cove = np.array([[1, 0.5], [0.5, 1]])
```

```python
n = 1000
sam = np.random.multivariate_normal(mu, cove, size = n)
```

```python
meand1 = np.mean(sam[:, 0])
meand2 = np.mean(sam[:, 1])
```

```python
muhat = np.mean(sam, axis = 0)
```

```python
muhat
```

```python
cen = sam - muhat
shat = (cen.T @ cen) / n
```

```python
eig1, eig2 = np.linalg.eigh(shat)
```

```python
eig1
```

```python
eig2
```

```python
from matplotlib.patches import Ellipse
```

```python
width = 2 * 2 * np.sqrt(eig1[1])
height = 2 * 2 * np.sqrt(eig1[0])
```

```python
ellipse = Ellipse(xy = muhat, width = width, height = height, 
                  angle = np.degrees(np.arctan2(eig2[1, 1], eig2[0,  1])), fill = False, edgecolor = 'red', linewidth = 2)
plt.gca().add_patch(ellipse)
plt.scatter(sam[:, 0], sam[:, 1], alpha=0.3)
```
