```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
np.random.seed(33550336)
```

```python
mu = 5
sigma = 2
n = 30
```

```python
from numpy.random import normal
```

```python
normal(mu, 4)
```

```python
se = 2/np.sqrt(30)
cmt = 0
```

```python
for _ in range(10000):
    sam = normal(mu, sigma, size = 30)
    me = 1.96 * se
    x = sam.mean()
    lowerb = x - me
    upperb = x + me
    if lowerb <= mu <= upperb:
        cmt += 1
```

```python
cmt
```

```python
n = 20
dat = normal(mu, sigma, n)
statabc =  []
```

```python
for _ in range(10000):
    x = np.random.choice(dat, size = len(dat), replace = True)
    statabc.append(np.mean(x))
```

```python
np.percentile(statabc, [2.5, 97.5])
```

```python
np.mean(dat)
```

```python
se = np.std(dat, ddof = 1)/np.sqrt(20)
```

```python
lowerb = np.mean(dat) - se * sp.stats.t.ppf(0.975, df = 19)
upperb = np.mean(dat) + se * sp.stats.t.ppf(0.975, df = 19)
```

```python
print(np.percentile(statabc, [2.5, 97.5]))
print(lowerb)
print(upperb)
```
