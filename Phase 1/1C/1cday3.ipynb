```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
from scipy.stats import norm
```

```python
m = 0
si = 1
```

```python
x = np.linspace(m - 4 * si, m + 4 * si, 500)
plt.plot(x, norm.pdf(x, loc = m, scale = si))
```

```python
plt.plot(x, norm.cdf(x, loc = m, scale = si))
```

```python
s = norm.rvs(loc = m, scale = si, size = 10000)
```

```python
np.mean(s)
```

```python
np.var(s, ddof = 1)
```

```python
from scipy.stats import uniform
np.random.seed(33550336)
```

```python
n = 100000
```

```python
sms = []
for i in range(10000):
    s = uniform.rvs(loc = 0,scale = 1, size = n)
    sms.append(np.mean(s))

std = 1/np.sqrt(12 * n)
x = np.linspace(0.5 - 4*std, 0.5 + 4*std, 500)
```

```python
plt.hist(sms, bins = 20, density = True)
plt.plot(x , norm.pdf(x, loc = 0.5, scale = std))
plt.show()
```
