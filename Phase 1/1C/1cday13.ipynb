```python
def pxf(k ,eta):
    h = 1 / sp.special.factorial(k)
    T = k
    A = np.exp(eta)
    return h * np.exp(eta * T - A)
```

```python
k = 5
lam = 2.3
eta = np.log(lam)
print(pxf(k ,eta))
print(sp.stats.poisson.pmf(k, lam))
```

```python
def gxf(x, mu, sigma2):
    e1 = mu / sigma2
    e2 = -1 / (2 * sigma2)
    t1 = x
    t2 = x ** 2
    h = 1 / np.sqrt(2 * np.pi)
    A = -e1 ** 2 / (4 * e2) - 0.5 * np.log(-2 * e2)
    return h * np.exp(e1 * t1 + e2 * t2 - A)
```

```python
x = 1.5
mu = 0.0
s2 = 2.0
print(gxf(x, mu, s2))
print(sp.stats.norm.pdf(x, mu, np.sqrt(s2)))
```

```python
def bxf(x, p):
    et = np.log(p / (1 - p))
    T = x
    h = 1
    A = np.log(1 + np.exp(et))
    return h * np.exp(et * T - A)
```

```python
x = 1
p = 0.3
print(bxf(x, p))
print(sp.stats.bernoulli.pmf(x, p))
```

```python
bet = np.array([0.5, -0.3])
```

```python
x = np.random.normal(loc = 0, scale = 1, size = (100, 2))
lam = np.exp(x @ bet)
y = np.random.poisson(lam)
```

```python
def nll(beta, X, y):
    lam = np.exp(X @ beta)
    ll = np.sum(y * (X @ beta) - lam)
    return -ll
```

```python
bi = np.array([0.0, 0.0])
res = sp.optimize.minimize(nll, bi, args = (x, y), method = 'BFGS')
bh = res.x 
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
x = np.random.normal(loc = 0, scale = 1, size = (100, 2))
bt = np.array([1.0, -0.5])
p = sigmoid(x @ bt)
y = np.random.binomial(1, p)
```

```python
res = sp.optimize.minimize(nll, bt, args = (x, y), method = 'BFGS')
bh = res.x 
```
