```python
t1c = 0
```

```python
for _ in range(10000):
    samp = np.random.normal(loc = 5, scale = 2, size = 25)
    sm = sp.stats.ttest_1samp(samp, 5.0).pvalue
    if sm < 0.05:
        t1c += 1
```
