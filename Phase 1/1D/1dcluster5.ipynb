```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
np.random.seed(33550336)
```

```python
x = np.array([[0, 0], [1, 1]])
y = np.array([-1, 1])
k = x @ x.T
```

```python
k
```

```python
bnd = [(0, None), (0, None)]
```

```python
def eq(alph):
    return np.sum(alph * y)
```

```python
def som(alph):
    t1 = np.sum(alph)
    sm = 0
    for i in range(2):
        for j in range(2):
            sm += alph[i] * alph[j] * y[i] * y[j] * k[i, j]
    t2 = sm * 0.5
    return -(t1 - t2)
```

```python
sm = {'type' : 'eq', 'fun' : eq}
```

```python
sm
```

```python
res = sp.optimize.minimize(som, x0=[0.1, 0.1], method='SLSQP', bounds=bnd, constraints=sm)
```

```python
alpha = res.x
```

```python
w = np.sum(alpha[:, None] * y[:, None] * x, axis = 0)
```

```python
b = y[1] - np.dot(w, x[1])
```
