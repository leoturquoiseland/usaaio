```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
np.random.seed(33550336)
```

```python
X = np.random.normal(loc = 2, size = (100, 2))
```

```python
np.shape(X)
```

```python
beta = [2, -3]
samp = 100
mn = 0
tsd = 0.001
ns = np.random.normal(mn, tsd, samp)
```

```python
y = X @ beta + ns
lam = 0.1
I = np.eye(2)
```

```python
br = np.linalg.inv(X.T @ X + lam * I) @ X.T @ y
```

```python
br
```

```python
bg = np.array([0.0, 0.0])
eta = 0.0001
```

```python
for _ in range(200):
    grad = -2 * X.T @ (y - X @ bg) + 2 * lam * bg
    bg = bg - eta * grad
```

```python
bt = np.array([1.0, -0.5])
x = np.random.normal(0, 1, (200, 2))
p = 1 / ( 1 + np.exp(-x @ bt))
y = np.random.binomial(1, p)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def lss(beta, x, y):
    p = sigmoid(x @ beta)
    return -np.sum(y * np.log(p) + (1 - y) * np.log(1 - p))
```

```python
def grad(beta, X, y):
    p = sigmoid(X @ beta)
    return X.T @ (p - y)
```

```python
beta = np.array([0.0, 0.0])
eta = 0.01
for _ in range(1000):
    beta = beta - eta * grad(beta, x, y)
```

```python
print("Loss at fitted:", lss(beta, x, y))
print("Loss at true:", lss(bt, x, y))
```
