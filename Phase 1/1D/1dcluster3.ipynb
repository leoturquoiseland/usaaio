```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
np.random.seed(33550336)
```

```python
bet = 0.9
```

```python
def f(x):
    return (x[0] - 2) ** 2 + 10 * (x[1] - 3) ** 2
```

```python
def gradf(x):
    return np.array([2 * (x[0] - 2), 20 * (x[1] - 3)])
```

```python
vs = []
xs = []
x = np.array([0.0, 0.0])
v = np.array([0.0, 0.0])
eta = 0.01
```

```python
for t in range(200):
    vt = bet * v + gradf(x)
    xt = x - eta * vt
    x = xt
    v = vt
    vs.append(v.copy())
    xs.append(x.copy())
```
