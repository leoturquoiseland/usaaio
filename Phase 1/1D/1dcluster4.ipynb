```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
np.random.seed(33550336)
```

```python
def f(x):
    return x[0] ** 2 + x[1] ** 2
```

```python
def eq(x):
    return x[0] + x[1] - 1
```

```python
cons = {'type': 'eq', 'fun': eq}
```

```python
res = sp.optimize.minimize(f, x0 = [0, 0], method = 'SLSQP', constraints = cons)
```
