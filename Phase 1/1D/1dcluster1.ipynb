```python
x = np.linspace(0, 20, 500)[1:]
y = np.linspace(-5, 9, 500)
```

```python
plt.plot(y, np.exp(y))
plt.plot(x, -np.log(x))
plt.plot(x, x ** 2)
plt.plot(x, x ** 3)
plt.plot(y, y**4  - 3 * y ** 3 + 2)
```
