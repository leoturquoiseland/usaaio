```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
np.random.seed(33550336)
```

```python
def gradf(x):
    return np.array([2 *(x[0] - 2) , 20 * (x[1] - 3)])
```

```python
def f(x):
    return(x[0] - 2) ** 2 + 10 * (x[1] - 3) ** 2
```

```python
l001 = np.zeros((200, 2))
l01 = np.zeros((200, 2))
l02 = np.zeros((200, 2))
l05 = np.zeros((200, 2))
```

```python
z001 = []
z01 = []
z02 = []
z05 = []
```

```python
x = np.array([0, 0])
eta = 0.01
T = 200
for _ in range(T):
    x = x - eta * gradf(x)
    if f(x) > 1e10:
        break
    z001.append(f(x))
```

```python
plt.plot(l05[:, 0])
plt.plot(l02[:, 0])
plt.plot(l01[:, 0])
plt.plot(l001[:, 0])
```

```python
z001[z001 == np.inf] = np.nan
z01[z01 == np.inf] = np.nan
z02[z02 == np.inf] = np.nan
z05[z05 == np.inf] = np.nan
```

```python
plt.plot(z001, label='η=0.01')
plt.plot(z01, label='η=0.1')
plt.plot(z02, label='η=0.2')
plt.plot(z05, label='η=0.5')
plt.yscale('log')
plt.ylim(1e-5, 1e10)  # cap the y-axis
plt.legend()
plt.show()
```
