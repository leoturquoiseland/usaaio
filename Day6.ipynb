```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

# ── LINE PLOT ────────────────────────────────────────────
x = np.linspace(0, 2*np.pi, 200)
plt.figure(figsize=(10, 5))

plt.plot(x, np.sin(x), label='sin(x)', color='#2563eb', linewidth=2)
plt.plot(x, np.cos(x), label='cos(x)', color='#dc2626', linewidth=2, linestyle='--')
plt.plot(x, np.tanh(x), label='tanh(x)', color='#16a34a', linewidth=2, linestyle=':')
plt.xlabel('x', fontsize=12); plt.ylabel('y', fontsize=12)
plt.title('Functions', fontsize=14, fontweight='bold')
plt.legend(fontsize=11); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

# ── HISTOGRAM ────────────────────────────────────────────
data = np.random.normal(0, 1, 5000)
plt.figure(figsize=(8, 4))
plt.hist(data, bins=50, density=True, color='steelblue', edgecolor='white', alpha=0.7)
x_pdf = np.linspace(-4, 4, 200)
plt.plot(x_pdf, np.exp(-x_pdf**2/2)/np.sqrt(2*np.pi), 'r-', lw=2, label='N(0,1) PDF')
plt.legend(); plt.title('Histogram + Theoretical PDF'); plt.show()
```
