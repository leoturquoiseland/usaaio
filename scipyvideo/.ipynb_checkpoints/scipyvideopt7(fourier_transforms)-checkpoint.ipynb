```python
from scipy.integrate import quad
from scipy.optimize import minimize
```

```python
def f(A):
    integrand = lambda t : 2 / (1 + A * np.sin(np.pi*t/10))) ** 2 + 5 * np.sqrt(1 + (np.pi * A/10) ** 2 * np.cos(np.pi * t/10) ** 2)
    return quad(integrand, 0, 10)[0]

minimize(f, 0.001).x
```
