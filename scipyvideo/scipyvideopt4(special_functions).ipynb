```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
from scipy.special import legendre
x = np.linspace(0, 1, 100)
plt.plot(x, legendre(6)(x))
plt.show()
```

```python
from scipy.special import jv
x = np.linspace(0, 10, 100)
plt.plot(x, jv(3, x))
plt.show()
```
