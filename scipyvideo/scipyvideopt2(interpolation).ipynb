```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
x = np.linspace(0, 10, 10)
y = x ** 2 * np.sin(x)
plt.scatter(x,y)
```

```python
from scipy.interpolate import interp1d
```

```python
f = interp1d(x, y, kind = 'cubic')
x_dense = np.linspace(0, 10, 100)
y_dense = f(x_dense)
```

```python
plt.plot(x_dense, y_dense)
```
