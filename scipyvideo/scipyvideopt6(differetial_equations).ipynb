```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
from scipy.integrate  import odeint
```

```python
def dvdt(v, t):
    return 3 * v ** 2 - 5

v0 = 0
```

```python
t = np.linspace(0, 1, 100)
sol = odeint(dvdt, v0, t)
sol
```

```python
sol.T[0]
```

```python
plt.plot(t, sol.T[0])
```

```python
def dSdx(S, x):
     y1, y2 = S
     return [y1 + y2 ** 2 + 3 * x,
            3 * y1 + y2 ** 3 - np.cos(x)]

y1_0 = 0
y2_0 = 0
S_0 = (y1_0, y2_0)
```

```python
x = np.linspace(0,1, 100)
sol = odeint(dSdx, S_0, x)
```

```python
sol
```

```python
sol.T
```

```python
y1 = sol.T[0]
y2 = sol.T[1]
```

```python
plt.plot(x, y1)
plt.plot(x, y2)
```

```python
def dSdt(S, t):
    theta, omega = S
    return[omega,
          np.sin(theta)]

theta0 = np.pi/4
omega0 = 0
S0 = (theta0, omega0)
```

```python
t = np.linspace(0, 20, 100)
sol = odeint(dSdt, S0, t)
theta, omega = sol.T
```

```python
plt.plot(t, theta)
plt.show()
```
